**#SEP PARAMETER**

```python
import pandas as pd

data = {
    "name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
    "age": [20, 22, 19, 21, 23],
    "marks": [85, 92, 78, 95, 67]
}

df = pd.DataFrame(data)

# Data ko TSV file mein save karo
df.to_csv("students.tsv", sep="\t", index=False)

print("TSV file created successfully!")
```

```python
import pandas as pd

df = pd.read_csv("students.tsv", sep="\t")

print(df)
```

```python
import pandas as pd
df=pd.read_csv("students.tsv",sep="\t")
print(df[["name","age"]])
```

**Index column**

```python
import pandas as pd

df = pd.read_csv(
    "students.tsv",
    sep="\t",
    index_col="name"
)

print(df)
               
               
```

**hearder parameter**

```python
import pandas as pd
df=pd.read_csv("students.tsv", sep='\t',header=1)
print(df)
```

**use cols parameter**

```python
import pandas as pd
df=pd.read_csv("students.tsv" ,sep='\t',usecols=["name","marks"])
print(df)
```

**skip rows**

```python
import pandas as pd
df=pd.read_csv("students.tsv",sep='\t',skiprows=[0,1])
print(df)
```
